# Data Cleaning and Preprocessing of Sentiment Dataset

## Objective
The aim of this project is to clean and preprocess a sentiment dataset by identifying missing values, removing duplicates, and standardizing inconsistent data formats using Python and Pandas.

```python
#Import Libraries
import pandas as pd
```

```python
#Load Dataset
df = pd.read_csv("Sentiment dataset.csv")
df.head()
```

## Load Dataset
The dataset is loaded into a Pandas DataFrame for inspection and preprocessing.

## Dataset Overview
Inspecting the dataset structure, data types, and non-null counts.

```python
df.info()
```

### Observation
The dataset contains 732 records and 15 columns. No missing values are immediately apparent from the non-null counts. The Timestamp column is stored as an object type and will require conversion to datetime format.

## Missing Value Analysis

```python
#Missing Values Check
df.isnull().sum()
```

### Observation
No missing values were found in any column of the dataset.

```python
#Duplicate Check
df.duplicated().sum()
```

### Observation
No duplicate records were identified in the dataset.

```python
#Remove Unnecessary Columns
df.drop(['Unnamed: 0', 'Unnamed: 0.1'], axis=1, inplace=True)
```

## Remove Unnecessary Columns

The dataset contains two unnecessary index columns (`Unnamed: 0` and `Unnamed: 0.1`) that were created during a previous CSV export. These columns do not contribute to the analysis and are removed to improve data quality.

```python
#Standardize Text Fields
cols = ['User', 'Platform', 'Country', 'Sentiment']

for col in cols:
    df[col] = df[col].str.strip()
```

## Standardize Text Fields

Text-based columns are standardized by removing leading and trailing spaces. This ensures consistency in categorical values and prevents issues during analysis.

```python
#Convert Timestamp
df['Timestamp'] = pd.to_datetime(df['Timestamp'])
```

## Convert Timestamp to Datetime Format

The `Timestamp` column is currently stored as a text (object) data type. It is converted to datetime format to enable proper date and time analysis.

```python
#Save Cleaned Dataset
df.to_csv('cleaned_sentiment_dataset.csv', index=False)
```

## Save the Cleaned Dataset

After completing all preprocessing steps, the cleaned dataset is exported as a new CSV file for future analysis and visualization tasks.

```python
#Final Preview
df.head()
```

## Preview of the Cleaned Dataset

The first few rows of the cleaned dataset are displayed to verify that the preprocessing steps were applied successfully.

# Conclusion

The sentiment dataset was successfully cleaned and preprocessed using Python and Pandas. Data quality checks confirmed that there were no missing values or duplicate records. Unnecessary index columns were removed, text fields were standardized, and the Timestamp column was converted to datetime format. The cleaned dataset is now ready for further analysis and visualization.
